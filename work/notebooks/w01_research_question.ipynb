## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*



```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Loading the starter dataset directly from the public repo
url = "https://raw.githubusercontent.com/Kavurubuvanesh/flyrank-ml-workspace/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Calculating the proof numbers
total_pages = len(df)
declining_pct = (df["trend_direction"] == "down").mean() * 100
high_traffic_declining = df[(df["trend_direction"] == "down") & ( df["impressions_90d"] > 10000)]

# Printing the justification
print(f"Total pages in starter sample: {total_pages}")
print(f"Percentage of the pages declining: { declining_pct:.1f}%")
print(f"High-traffic pages (>10k impressions) activity losing traffic: {len(high_traffic_declining)}")
print("\nConlusion: With over half the sample declining, and hundreds of high-value pages at risk, automating the detection process is highly justified.")

```
